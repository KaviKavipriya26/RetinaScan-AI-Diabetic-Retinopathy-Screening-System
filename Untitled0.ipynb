```python

import os
import zipfile
from pathlib import Path
from collections import Counter

ZIP_PATH = "/content/sample_data/colored_images.zip"
EXTRACT_DIR = "/content/disease_dataset"

if not os.path.exists(ZIP_PATH):
    raise FileNotFoundError(
        f"Cannot find {ZIP_PATH}. Upload the ZIP to Colab first."
    )

os.makedirs(EXTRACT_DIR, exist_ok=True)

with zipfile.ZipFile(ZIP_PATH) as archive:
    bad_file = archive.testzip()
    if bad_file:
        raise ValueError(f"Corrupt ZIP file entry: {bad_file}")
    archive.extractall(EXTRACT_DIR)

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def count_images(directory):
    return sum(
        1 for path in Path(directory).rglob("*")
        if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
    )

# Locate the directory containing the disease class folders.
candidates = []

for current, subdirs, _ in os.walk(EXTRACT_DIR):
    valid_subdirs = [
        name for name in subdirs
        if count_images(os.path.join(current, name)) > 0
    ]
    if len(valid_subdirs) >= 2:
        candidates.append((len(valid_subdirs), current))

if not candidates:
    raise RuntimeError(
        "Could not locate at least two image class folders. "
        "Inspect EXTRACT_DIR and set DATASET_DIR manually."
    )

candidates.sort(key=lambda item: item[0], reverse=True)
DATASET_DIR = candidates[0][1]

class_names = sorted([
    p.name for p in Path(DATASET_DIR).iterdir()
    if p.is_dir() and count_images(p) > 0
])

class_counts = {
    name: count_images(Path(DATASET_DIR) / name)
    for name in class_names
}

print("Dataset directory:", DATASET_DIR)
print("Number of classes:", len(class_names))
print("Total images:", sum(class_counts.values()))

for name, count in class_counts.items():
    print(f"{name}: {count}")

assert len(class_names) >= 2, "At least two classes are required."
```

```python

import numpy as np
from sklearn.model_selection import train_test_split

SEED = 42
IMG_SIZE = 224
BATCH_SIZE = 32
NUM_CLASSES = len(class_names)

class_to_index = {
    name: i for i, name in enumerate(class_names)
}

image_paths = []
image_labels = []

for class_name in class_names:
    folder = Path(DATASET_DIR) / class_name

    for path in folder.rglob("*"):
        if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS:
            image_paths.append(str(path))
            image_labels.append(class_to_index[class_name])

image_paths = np.asarray(image_paths)
image_labels = np.asarray(image_labels)

counts = Counter(image_labels.tolist())

if min(counts.values()) < 3:
    raise ValueError(
        "At least 3 images per class are needed for this split. "
        "More images are recommended."
    )

train_paths, remaining_paths, train_labels, remaining_labels = (
    train_test_split(
        image_paths,
        image_labels,
        test_size=0.30,
        random_state=SEED,
        stratify=image_labels
    )
)

val_paths, test_paths, val_labels, test_labels = train_test_split(
    remaining_paths,
    remaining_labels,
    test_size=0.50,
    random_state=SEED,
    stratify=remaining_labels
)

print("Training:", len(train_paths))
print("Validation:", len(val_paths))
print("Testing:", len(test_paths))
print("Classes:", class_names)
```

```python

import tensorflow as tf

tf.keras.utils.set_random_seed(SEED)

AUTOTUNE = tf.data.AUTOTUNE

def load_image(path, label):
    image_bytes = tf.io.read_file(path)

    image = tf.io.decode_image(
        image_bytes,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = tf.cast(image, tf.float32)

    label = tf.one_hot(label, NUM_CLASSES)
    return image, label

def create_dataset(paths, labels, training=False):
    dataset = tf.data.Dataset.from_tensor_slices((paths, labels))

    if training:
        dataset = dataset.shuffle(
            len(paths),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)
    return dataset.prefetch(AUTOTUNE)

train_ds = create_dataset(train_paths, train_labels, training=True)
val_ds = create_dataset(val_paths, val_labels)
test_ds = create_dataset(test_paths, test_labels)

print("Training pipeline ready.")
```

```python

from tensorflow import keras
from tensorflow.keras import layers

OUTPUT_DIR = "/content/disease_model_output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

BEST_MODEL_PATH = os.path.join(
    OUTPUT_DIR, "best_disease_model.keras"
)

augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomZoom(0.10),
    layers.RandomContrast(0.10),
], name="augmentation")

base_model = keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)
base_model.trainable = False

inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = augmentation(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.30)(x)
x = layers.Dense(256, activation="relu")(x)
x = layers.Dropout(0.20)(x)
outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax",
    name="disease_classes"
)(x)

model = keras.Model(inputs, outputs)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python

callbacks_stage1 = [
    keras.callbacks.ModelCheckpoint(
        BEST_MODEL_PATH,
        monitor="val_accuracy",
        mode="max",
        save_best_only=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-7,
        verbose=1
    )
]

history_stage1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=35,
    callbacks=callbacks_stage1,
    verbose=1
)

print("Stage 1 finished.")
print("Best checkpoint:", BEST_MODEL_PATH)
print(
    "Best validation accuracy:",
    f"{max(history_stage1.history['val_accuracy']) * 100:.2f}%"
)
```

```python

# Begin from the best model produced in Stage 1.
model = keras.models.load_model(BEST_MODEL_PATH)

base_model = model.get_layer("efficientnetb0")
base_model.trainable = True

# Freeze most layers and BatchNormalization layers.
for layer in base_model.layers:
    layer.trainable = False

for layer in base_model.layers[-50:]:
    if not isinstance(layer, layers.BatchNormalization):
        layer.trainable = True

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

callbacks_stage2 = [
    keras.callbacks.ModelCheckpoint(
        BEST_MODEL_PATH,
        monitor="val_accuracy",
        mode="max",
        save_best_only=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-8,
        verbose=1
    )
]

history_stage2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=35,
    callbacks=callbacks_stage2,
    verbose=1
)

# Use the best validation-accuracy checkpoint from both stages.
model = keras.models.load_model(BEST_MODEL_PATH)

print("Stage 2 finished.")
print("Best checkpoint restored.")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
import matplotlib.pyplot as plt
import seaborn as sns

test_loss, test_accuracy_keras = model.evaluate(test_ds, verbose=1)

probabilities = model.predict(test_ds, verbose=1)
predicted_labels = np.argmax(probabilities, axis=1)

true_labels = np.concatenate([
    np.argmax(labels.numpy(), axis=1)
    for _, labels in test_ds
])

test_accuracy = accuracy_score(true_labels, predicted_labels)

print("\n" + "=" * 45)
print(f"FINAL TEST ACCURACY: {test_accuracy * 100:.2f}%")
print(f"TEST LOSS: {test_loss:.4f}")
print("=" * 45)

if test_accuracy >= 0.90:
    print("TARGET ACHIEVED: Test accuracy is at least 90%.")
else:
    print("TARGET NOT YET ACHIEVED: Test accuracy is below 90%.")
    print("Inspect the per-class metrics and confusion matrix.")

report = classification_report(
    true_labels,
    predicted_labels,
    labels=np.arange(NUM_CLASSES),
    target_names=class_names,
    zero_division=0
)

print("\nPer-class performance:")
print(report)

cm = confusion_matrix(
    true_labels,
    predicted_labels,
    labels=np.arange(NUM_CLASSES)
)

plt.figure(figsize=(max(10, NUM_CLASSES), max(8, NUM_CLASSES * 0.7)))
sns.heatmap(
    cm,
    annot=NUM_CLASSES <= 15,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)
plt.xlabel("Predicted disease")
plt.ylabel("Actual disease")
plt.title("Confusion Matrix — Held-out Test Set")
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()

cm_path = os.path.join(OUTPUT_DIR, "confusion_matrix.png")
plt.savefig(cm_path, dpi=150, bbox_inches="tight")
plt.show()

with open(
    os.path.join(OUTPUT_DIR, "evaluation_report.txt"),
    "w",
    encoding="utf-8"
) as f:
    f.write(f"Test accuracy: {test_accuracy:.6f}\n")
    f.write(f"Test loss: {test_loss:.6f}\n\n")
    f.write(report)
```

```python

import json
import os
import shutil
from pathlib import Path
from google.colab import files
from tensorflow import keras

OUTPUT_DIR = "/content/disease_model_output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

FINAL_MODEL_PATH = os.path.join(
    OUTPUT_DIR, "disease_detection_model.keras"
)

CLASS_NAMES_PATH = os.path.join(
    OUTPUT_DIR, "class_names.json"
)

METRICS_PATH = os.path.join(
    OUTPUT_DIR, "test_metrics.json"
)

REPORT_PATH = os.path.join(
    OUTPUT_DIR, "evaluation_report.txt"
)

# 1. Save the best selected model.
model.save(FINAL_MODEL_PATH)

# 2. Reload it to verify the saved model.
reloaded_model = keras.models.load_model(
    FINAL_MODEL_PATH,
    compile=False
)

if reloaded_model.input_shape[1:] != (224, 224, 3):
    raise RuntimeError(
        f"Unexpected model input: {reloaded_model.input_shape}"
    )

if reloaded_model.output_shape[-1] != NUM_CLASSES:
    raise RuntimeError(
        f"Unexpected model output: {reloaded_model.output_shape}"
    )

# 3. Export the exact output-index-to-class mapping.
with open(CLASS_NAMES_PATH, "w", encoding="utf-8") as f:
    json.dump(class_names, f, indent=2)

# 4. Export measured evaluation metrics.
with open(METRICS_PATH, "w", encoding="utf-8") as f:
    json.dump(
        {
            "test_accuracy": float(test_accuracy),
            "test_accuracy_percent": float(test_accuracy * 100),
            "test_loss": float(test_loss),
            "num_classes": int(NUM_CLASSES),
            "class_names": class_names
        },
        f,
        indent=2
    )

# 5. Save the classification report.
with open(REPORT_PATH, "w", encoding="utf-8") as f:
    f.write(f"Test accuracy: {test_accuracy:.6f}\n")
    f.write(f"Test loss: {test_loss:.6f}\n\n")
    f.write(report)

# 6. Package the files for the backend.
ZIP_OUTPUT = shutil.make_archive(
    "/content/disease_detection_model_package",
    "zip",
    root_dir=OUTPUT_DIR
)

print("Model:", FINAL_MODEL_PATH)
print("Class labels:", class_names)
print(f"Test accuracy: {test_accuracy * 100:.2f}%")
print("Package:", ZIP_OUTPUT)

files.download(ZIP_OUTPUT)
```
